## Streaming and Batch

## Streaming

Most models can stream their output content while it is being generated. by displaying output progressively, streaming significantly improves user experience, particularly for longer responses. Calling stream() returns an iterator that yields output chunks as they are produced. You can use a loop to process each chunk in real-time:

```python
## here we need to wait till the entire output is generated

model.invoke("Write me a 200 words paragraph on Artificial Intelligent")
```

```python
## use the stream() method which will send the output as chunks, while it is being generated and we are just prining the chunks. the flush() will basically make sure that forces Python to immediately write that output to your terminal, instead of holding it in an internal buffer and writing it later in a batch.

for chunk in model.stream("Write me a 200 words paragraph on Artificial Intelligent"):
  print(chunk.text, end="|", flush=True)
```

## Batch

Batching is a collection of independent/parallel requests to a model can significantly improve performance and reduce costs, as the processing can be done in parallel:

```python
response = model.batch([
    "Why do parrots have colourful feathers?",
    "How do airplanes fly?",
    "What is quantum compouting"
  ],
  config={
    'max_concurrency':5, # limit to 5 parallel calls
  }
)
response
```
