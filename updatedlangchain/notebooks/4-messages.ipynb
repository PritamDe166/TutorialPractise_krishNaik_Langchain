## Messages

Messages are the fundamental unit of context for the models in LangChain. They represent the input and output of the models, carrying both the content and the metadata needed to represent the state of conversation when interacting with an LLM.
messages are objects that contain:
- Role : Indentifies the message type (system, user)
- Content : Represents the actual content of the message (like text, documents, audio, images etc.)
- Metadata :  Optional fields such as response information, messageIds and token usage

Langchain provides and standard message type that works across all providers, ensuring consistant behaviour regardless of the model being called.

```python
import os
from dotenv import load_dotenv
import anthropic
from langchain.tools import tool
from langchain.chat_models import init_chat_model
from langchain.agents import create_agent
from langchain_core.messages import HumanMessage
from pprint import pprint

load_dotenv()

os.environ["ANTHROPIC_API_KEY"] = os.getenv("ANTHROPIC_API_KEY")

model = init_chat_model("claude-sonnet-5")

model.invoke("Tell me 5 facts about India")
```

### Text Prompts

Use text prompts when :
- You have a single, standalone request.
- You dont need convesation history.
- You want minimal code complexity.

```python
model.invoke("What is the capital of Japan?")
```

### Message Prompts

Alternatively you can pass in a list of messages to the model by providing a list if message objects.

Message Types:
- System Message: Tell the model how to behave and provide context for interactions.
- Human Message: Represents the user input and interactions with the model
- AI message: Represents the responses generated by the model, including the text, tool calls and the metadata
- Tool message: Represents the output of the local tools. 

```python
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage, ToolMessage

messages = [
  SystemMessage("You are a helpful assistant"),
  HumanMessage("What is the total travel time via various transportation from India to Srilanka?")
]

response = model.invoke(messages)
response.content
```

```python
## Detailed information to the llm through SystemMessage

system_message01 = SystemMessage("""You are a senior python developer specializing in web frameworks.
Always provide code examples and reasoning with your responses.
be concise but thorough in your explainations""")

messages01 = [system_message01,
              HumanMessage("How can we create a rest API?")]

reponse01 = model.invoke(messages01)
print(reponse01.content)
```

```python
## message metadata

human_message01 = HumanMessage(
  content="hello",
  name="Alice",
  id="msg_123"
)

response02 = model.invoke([human_message01])
response02
```

```python
## Creating AI message manually

messages03 = [
  SystemMessage("You are a helpful assistant"),
  HumanMessage("Can You help me?"),
  AIMessage("How may i help you"),
  HumanMessage("What is 2+2?")
]

response03 = model.invoke(messages03)
print(response03.content)
```

```python
response03.usage_metadata
```

```python
ai_message04 = AIMessage(
  content=[],
  tool_calls=[{
    "name":"get_weather",
    "args":{"location":"San-Francisco"},
    "id":"call_123"
  }]
)

weather_result ="Sunny, 72F"

tool_message = ToolMessage(
  content=weather_result,
  tool_call_id="call_123"
)

messages04 = [
  HumanMessage("What is the weather like in San Francisco?"),
  ai_message04,
  tool_message
]

response04 = model.invoke(messages04)
print(response04.content)
```
